```python
from pathlib import Path
import csv
import matplotlib.pyplot as plt

# eval_video_path = Path("runs/071026_212603/eval_videos")
# eval_video_path = Path("runs/071026_203708/eval_videos")
# eval_video_path = Path("runs/081026_152041/eval_videos")
eval_video_path = Path("runs/081026_161035/eval_videos")
eval_video_path = Path("../runs/101026_105636/eval_videos")
eval_video_path = Path("/home/liyouzhou/LearningFromHumanVideo/runs/121026_095026/eval_videos")
eval_video_path = Path("/home/liyouzhou/LearningFromHumanVideo/runs/121026_120213/eval_videos")

step_folders = [f for f in eval_video_path.iterdir()]
step_folders.sort()

fg, axs = plt.subplots(max(len(step_folders),2), 3, figsize=(15, 5*len(step_folders)))
plt.tight_layout()

for i, step_folder in enumerate(step_folders):
    step_number = int(str(step_folder).split("_")[-1])
    csv_file = step_folder / "results.csv"
    with open(csv_file, "r") as f:
        reader = csv.reader(f)
        data = list(reader)

        for episode in range(max(int(row[0]) for row in data[1:]) + 1):
            episode_rows = [row for row in data[1:] if int(row[0]) == episode]

            episode_imitation_rewards = [float(row[2]) for row in episode_rows]
            episode_gripper_rewards = [float(row[3]) for row in episode_rows]
            episode_z_rewards = [float(row[4]) for row in episode_rows]

            total_reward = sum(episode_imitation_rewards) + sum(episode_gripper_rewards) + sum(episode_z_rewards)

            ax = axs[i, episode]
    
            ax.plot(episode_imitation_rewards, label="Imitation Reward")
            ax.plot(episode_gripper_rewards, label="Gripper Reward")
            ax.plot(episode_z_rewards, label="Z Reward")
            ax.set_title(f"Step {step_number} Episode {episode} Total Reward {total_reward:.2f}")
            ax.legend()
```
